```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
```

```python
df = pd.DataFrame({'Gender' : ['Male','Female','Male','Female','Male','Female'] ,
                    'Region' :['East','West','South','North','Central','East'],
                    'Rented_Ownhose':['Rent','OwnHouse','Rent','OwnHouse','Rent','OwnHouse'],
                   'Occupration' : ['It Employee' , 'GOVT Officer' , 'Student_intern' , 'Doctor','Teacher','Salesman'],
                   'Education' : ['BTECH','Inter','PG','MBBS','phd','Undergraduate']
                     })


df
```

```python
# Gender and region columns are nominal columns
# Rented_Ownhouse , Occupation and Education are Odinal columns
```

```python
# use onehot encoding for Nominal columns
# use Odinal encoding for odinal columns
```

```python
from sklearn.preprocessing import OneHotEncoder
one_hot_obj = OneHotEncoder(drop='first')

one_hot_obj.fit(df[['Gender','Region']])

result = one_hot_obj.transform(df[['Gender','Region']])
```

```python
one_hot_obj.categories_
```

```python
one_hot_obj.get_feature_names_out()
```

```python
result = result.toarray()
result
```

```python
t = pd.DataFrame(data = result , columns=one_hot_obj.get_feature_names_out() , index = ['r1','r2','r3','r4','r5','r6'])
t
```

```python
df.index
```

```python
t.index
```

```python
df.reset_index(drop=True , inplace=True)
```

```python
df.index
```

```python
t.reset_index(drop=True , inplace=True)

t.index
```

```python
df = pd.concat([df , t] ,axis=1)

df
```

```python
df = df.drop(['Gender' , 'Region'] , axis = 1)
df
```

```python
df.columns
```

```python
# Odinal Technique

from sklearn.preprocessing import OrdinalEncoder
odinal_obj = OrdinalEncoder()

odinal_obj.fit(df[['Rented_Ownhose', 'Occupration', 'Education']])

result_1 = odinal_obj.transform(df[['Rented_Ownhose', 'Occupration', 'Education']])

result_1
```

```python
odinal_obj.categories_
```

```python
odinal_obj.get_feature_names_out() + '_res'
```

```python
p = pd.DataFrame(data = result_1 , columns=odinal_obj.get_feature_names_out() + '_res')
p
```

```python
df = pd.concat([df , p],axis=1)
```
