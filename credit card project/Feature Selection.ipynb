**Correlation with Hypothesis Testing**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
import seaborn as sns

df = pd.read_csv("/content/Titanic-Dataset.csv")
df = df.dropna()
df_num_cols = df.select_dtypes(exclude = 'object')
df_num_cols
```

```python
X = df_num_cols.drop(['Survived'],axis=1)
y = df_num_cols['Survived']
```

```python
from scipy.stats import pearsonr
```

```python
corr_pvalues = []

for i in X.columns:
  values = pearsonr(X[i] , y)
  corr_pvalues.append(values)

```

```python
corr_pvalues
```

```python
corr_pvalues = np.array(corr_pvalues)
```

```python
corr_pvalues
```

```python
p_values = corr_pvalues[ : , 1]
```

```python
X.columns
```

```python
p_values
```

```python
f = pd.Series(p_values , index = X.columns )
f
```

```python
c = 0
for i in f:
  if i > 0.05:
    print(X.columns[c] , i)
  # else:
  #   print(X.columns[c] , i)
  c = c + 1
```

```python
plt.figure(figsize = (5,3))
plt.title("Hypothesis Testing")

plt.xlabel("column Names")
plt.ylabel("P_values for Each Independent column")

plt.bar(X.columns , p_values)

plt.show()
```
